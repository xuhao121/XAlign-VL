```python
import pandas as pd
import gzip
import os

# 本地基础路径
base_path = '.Data/Raw Data'

# 映射各个文件的相对路径 (已更新为 
image_list_file = os.path.join(base_path, 'train.txt') 
en_file = os.path.join(base_path, 'train.en.gz')
de_file = os.path.join(base_path, 'train.de.gz')
fr_file = os.path.join(base_path, 'train.fr.gz')
cs_file = os.path.join(base_path, 'train.cs.gz')

# 编写一个智能读取函数，自动识别并处理 .gz 压缩文件
def read_lines(filepath):
    if not os.path.exists(filepath):
        raise FileNotFoundError(f"Cannot find files: {filepath}")
        
    if filepath.endswith('.gz'):
        with gzip.open(filepath, 'rt', encoding='utf-8') as f:
            return [line.strip() for line in f]
    else:
        with open(filepath, 'r', encoding='utf-8') as f:
            return [line.strip() for line in f]

try:
    print("Reading files...")
    images = read_lines(image_list_file)
    en_captions = read_lines(en_file)
    de_captions = read_lines(de_file)
    fr_captions = read_lines(fr_file)
    cs_captions = read_lines(cs_file)
    
    # 检查所有文件的行数是否完全一致
    assert len(images) == len(en_captions) == len(de_captions) == len(fr_captions) == len(cs_captions), "文件行数不匹配，数据没有正确对齐！"

    # 组合成 DataFrame
    df = pd.DataFrame({
        'image_id': images,
        'caption_en': en_captions,
        'caption_de': de_captions,
        'caption_fr': fr_captions,
        'caption_cs': cs_captions
    })

    # 按照项目提案，抽取 500 个样本
    sample_size = 500
    df_sampled = df.sample(n=sample_size, random_state=42)

    # 保存到你的数据文件夹中
    output_file = os.path.join(base_path, 'multi30k_sample_500.csv')
    df_sampled.to_csv(output_file, index=False, encoding='utf-8')
    
    print(f"Successfully data cleaning and alignment.")
    print(f"Have saved {sample_size} images and multipul descriptions to: \n{output_file}")

except Exception as e:
    print(f"❌ 运行出错: {e}")
```
